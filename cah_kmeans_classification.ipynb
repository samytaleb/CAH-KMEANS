```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import scatter_matrix
```

# **Read data**

```python
# Import des données
data_path = "/drive_path/AnaDo_JeuDonnees_TemperatFrance.xlsx"
```

```python
data = pd.read_excel(data_path)
data = data.set_index('Unnamed: 0')
data
```

```python
print(data.keys())
```

```python
X = data.values[:,:12]
print(X.shape)
print(X)
```

## **Data standardization**

```python
# On centre et on réduit
import sklearn
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_cr = sc.fit_transform(X)
print(X_cr)
```

# **CAH classification**

```python
# Library for CAH
from matplotlib import pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage, fcluster

# Generate linkage matrix
Z = linkage(X_cr,method='ward',metric='euclidean')
```

```python
print(Z)

```

```python
n = Z.shape[0] + 1
nb_clusters = np.arange(n-1, 0, -1)  # from n to 2 clusters

# Heights of the merges
merge_heights = Z[:, 2]  # Ward distance (proportional to ΔI_intra)

# Histogram: each bar corresponds to a merge
plt.figure(figsize=(10, 5))
plt.bar(nb_clusters, merge_heights, width=0.7, color='skyblue', edgecolor='k')
plt.xlabel("Number of clusters")
plt.ylabel("Merge height (Z[:,2])")
plt.title("Evolution of merge distance with number of clusters")
plt.grid(axis='y')
plt.show()
```

```python
# Show dendrogram
plt.figure(figsize=(15, 10))
plt.title("CAH")
dendrogram(Z,labels=data.index, color_threshold = 5.5)
plt.show()
```

```python
groupes_cah = fcluster(Z,t=5.5,criterion='distance')
print(groupes_cah)
```

```python
# Create a DataFrame with the cities and their class labels
df = pd.DataFrame({'City': data.index, 'Cluster': groupes_cah})
df
```

```python
# Sort in increasing order
df.sort_values(by=['Cluster'])
```

# **K-MEANS**

```python
# Library for cluster evaluation
from sklearn import cluster
from sklearn import metrics

# Using the "silhouette" metric
# Vary the number of clusters from 2 to 9
res = []
for k in range(2,10):
    km = cluster.KMeans(n_clusters=k).fit(X_cr)
    res.append(metrics.silhouette_score(X_cr,km.labels_))
print(res)
# graphique
plt.title("Silhouette Scores for Different Numbers of Clusters")
plt.xlabel("Number of clusters")
plt.ylabel("Silhouette Score")
plt.grid(True)
plt.plot(np.arange(2,10,1),res)
plt.show()
```

```python
# Initialize an empty list to store within-cluster inertia
inertias = []

# Vary the number of clusters from 1 to 10
for k in range(1, 11):
    # Perform K-means clustering
    kmeans = cluster.KMeans(n_clusters=k).fit(X_cr)
    # Append the within-cluster dispersion (inertia) to the list
    inertias.append(kmeans.inertia_)

# Plot the elbow graph
plt.plot(range(1, 11), inertias, marker='o')
plt.xlabel('Number of clusters')
plt.ylabel('Number of clusters')
plt.title("Elbow Method")
plt.grid(True)
plt.show()
```

```python
kmeans = cluster.KMeans(n_clusters=3)  # We keep n_clusters = 3
kmeans.fit(X_cr)
```

```python
# Create a DataFrame with the cheeses and their class labels
df2 = pd.DataFrame({'Ville':data.index, 'Cluster':kmeans.labels_})
df2.sort_values(by=['Cluster'])
```

```python
# Mapping to the CAH groups
pd.crosstab(groupes_cah,kmeans.labels_)
```

# **PCA**

```python
from sklearn.decomposition import PCA

# Instanciation of PCA
pca = PCA(n_components = 2)
# The PCA is performed, and the factor scores are retrieved in the matrix 'coord'.
coord = pca.fit_transform(X_cr)
print(coord)
```

```python
# Retrieve the eigenvalues / variance / inertia.
# The variance is estimated with n-1 degrees of freedom.
variance = pca.explained_variance_

# An adjustment is necessary to obtain the variance with n degrees of freedom.
eigval = variance*(len(X_cr)-1)/len(X_cr)
print('Eigenvalues (Variances of principal components) are :')
for index, lambda_k in enumerate(eigval):
  print(f'Variance of principal component PC{index+1}: Lambda_k = {lambda_k}')
```

```python
# Eigenvalue representation ratio or the percentage of inertia explained by each axis
np.set_printoptions(suppress=True)
print(pca.explained_variance_ratio_*100)
```

```python
# Creation of the graph
fig, ax = plt.subplots(figsize=(12,12))
ax.set_xlim(-7,7)
ax.set_ylim(-7,7)

for couleur,k in zip(['red','blue','lawngreen'],[0,1,2]):
    plt.scatter(coord[kmeans.labels_==k,0],coord[kmeans.labels_==k,1],c=couleur,label = f'Cluster_{k}')

# Placement of the observation labels
for i in range(len(X_cr)):
     plt.annotate(data.index[i],(coord[i,0],coord[i,1]))
# Axes
plt.plot([-7,7],[0,0], color='silver', linestyle='-', linewidth=1)
plt.plot([0,0],[-7,7], color='silver', linestyle='-', linewidth=1)

# Légende
plt.legend()

plt.show()
```
